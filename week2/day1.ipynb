## And now for some fun - an adversarial conversation between Chatbots..

You're already familar with prompts being organized into lists like:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "user prompt here"}
]
```

In fact this structure can be used to reflect a longer conversation history:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "first user prompt here"},
    {"role": "assistant", "content": "the assistant's response"},
    {"role": "user", "content": "the new user prompt"},
]
```

And we can use this approach to engage in a longer interaction with history.

````python
# Let's make a conversation between GPT-4.1-mini and Claude-3.5-haiku
# We're using cheap versions of models so the costs will be minimal

gpt_model = "gpt-4.1-mini"
claude_model = "anthropic/claude-3.5-haiku"

gpt_system = "You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

claude_system = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

gpt_messages = ["Hi there"]
claude_messages = ["Hi"]
````

````python
def call_gpt():
    messages = [{"role": "system", "content": gpt_system}]
    for gpt, claude in zip(gpt_messages, claude_messages):
        messages.append({"role": "assistant", "content": gpt})
        messages.append({"role": "user", "content": claude})
    response = openrouter.chat.completions.create(model=gpt_model, messages=messages)
    return response.choices[0].message.content
````

````python
call_gpt()
````

````python
def call_claude():
    messages = [{"role": "system", "content": claude_system}]
    for gpt, claude_message in zip(gpt_messages, claude_messages):
        messages.append({"role": "user", "content": gpt})
        messages.append({"role": "assistant", "content": claude_message})
    messages.append({"role": "user", "content": gpt_messages[-1]})
    response = openrouter.chat.completions.create(model=claude_model, messages=messages)
    return response.choices[0].message.content
````

````python
call_claude()
````

````python
gpt_messages = ["Hi there"]
claude_messages = ["Hi"]

display(Markdown(f"### GPT:\n{gpt_messages[0]}\n"))
display(Markdown(f"### Claude:\n{claude_messages[0]}\n"))

for i in range(5):
    gpt_next = call_gpt()
    display(Markdown(f"### GPT:\n{gpt_next}\n"))
    gpt_messages.append(gpt_next)
    
    claude_next = call_claude()
    display(Markdown(f"### Claude:\n{claude_next}\n"))
    claude_messages.append(claude_next)
````

````python
from IPython.display import display, Markdown
from openai import OpenAI

# --- Setup ---
openrouter = OpenAI(
    base_url=base_url,
    api_key=openrouter_api_key
)

# --- Example system prompts (yours can be different) ---
system_prompt_alex = """
You are Alex — an aggressively anti-vegan person who believes eating meat is natural, healthy, and part of human tradition.
You see veganism as unnecessary and often hypocritical.
You like to challenge vegans on their claims, use sarcasm, and emphasize freedom of choice, taste, and evolution.
You often accuse vegans of being overly emotional or unrealistic.
You are in a conversation with Charlie (a passionate vegan) and Blake (neutral, likes humor and avoids conflict).
You respond sharply and confidently, not holding back your opinions.
"""

user_prompt_alex = """
You are Alex, in conversation with Charlie and Blake.
Respond naturally to the conversation history, expressing your strong pro-meat stance.
Keep your tone assertive, sometimes sarcastic, and directly challenge vegan claims.
The conversation so far is as follows:
"""


# Example placeholders (define your own)
system_prompt_charlie = """
You are Charlie — a passionate and confrontational vegan.
You believe veganism is morally and scientifically superior to eating meat.
You argue that meat consumption causes animal suffering, environmental destruction, and health issues.
You often confront others for eating animal products and try to persuade them to switch to plant-based diets.
You are unafraid of debate and see yourself as fighting for justice and truth.
You are in a conversation with Alex (an anti-vegan) and Blake (neutral, humorous, dislikes conflict).
Ask pointed questions and challenge non-vegan logic, using facts, passion, and moral conviction.
"""

user_prompt_charlie = """
You are Charlie, in conversation with Alex and Blake.
Respond naturally to the conversation history, defending veganism strongly and confidently.
Ask sharp questions, counter meat-based arguments, and highlight ethical and environmental reasoning.
The conversation so far is as follows:
"""

system_prompt_blake = """
You are Blake — an undecided person caught between a heated vegan vs. meat-eating debate.
You don’t like conflict or tension, and when arguments get intense, you often crack light jokes or change the subject to cool things down.
You might occasionally ask simple or funny questions to both sides.
You haven’t decided whether veganism is good or bad — you just want everyone to relax.
You are in a conversation with Alex (aggressively anti-vegan) and Charlie (aggressively pro-vegan).
Your main goal is to ease the tension and keep things friendly, even when others get heated.
"""

user_prompt_blake = """
You are Blake, in conversation with Alex and Charlie.
Respond naturally to the conversation history, making lighthearted or humorous comments when things get tense.
You can ask silly or genuine questions, or try to steer the topic away from arguments.
Keep your tone relaxed and funny — you’re the comic relief.
The conversation so far is as follows:
"""


# --- Helper functions ---
def create_messages(role, conversation):
    """Format conversation history and return messages for a given role."""
    if role == "alex":
        system_prompt = system_prompt_alex
        user_prompt = user_prompt_alex
    elif role == "blake":
        system_prompt = system_prompt_blake
        user_prompt = user_prompt_blake
    elif role == "charlie":
        system_prompt = system_prompt_charlie
        user_prompt = user_prompt_charlie
    else:
        raise ValueError(f"Unknown role: {role}")

    # Build conversation history as dialogue text
    conversation_text = "\n".join(
        f"{msg['role'].capitalize()}: {msg['response']}" for msg in conversation
    )

    # Final user content
    full_user_prompt = f"{user_prompt}\n{conversation_text}\n"

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": full_user_prompt},
    ]
    return messages


def create_conversation(conversation, role, response):
    """Append a new turn to the conversation list."""
    conversation.append({"role": role, "response": response})
    return conversation


def conversation_loop(role, conversation):
    """Generate one response for the given role and display it."""
    response = openrouter.chat.completions.create(
        model="gpt-4.1-mini",
        messages=create_messages(role, conversation),
    )
    response = response.choices[0].message.content
    display(Markdown(response))

    return create_conversation(conversation, role, response)


# --- Running a sample conversation ---
roles = ["alex", "charlie", "blake"]
conversation = []

for _ in range(3):  # number of dialogue rounds
    for role in roles:
        conversation = conversation_loop(role, conversation)
````
